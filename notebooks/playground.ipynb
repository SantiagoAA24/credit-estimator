```python
rentabilidad_efectiva_anual = 0.1
inversion_mensual = 1000000
numero_de_anios = 20

rentabilidad_efectiva_mensual = (1 + rentabilidad_efectiva_anual) ** (1 / 12) - 1
numero_de_meses = numero_de_anios * 12
inversion_total = numero_de_meses * inversion_mensual

print(f'Rentabilidad E.A.: {rentabilidad_efectiva_anual * 100:.0f}%')
print(f'Rentabilidad E.M.: {rentabilidad_efectiva_mensual * 100:.0f}%')
print(f'Periodo de inversión [años]: {numero_de_anios}')
print(f'Periodo de inversión [meses]: {numero_de_meses}')

print("\n" + "-" * 50)
print("Si se invite cada mes la misma cantidad")
print("-" * 50)
print(f'Inversión mensual: ${inversion_mensual:,.0f}')
inversion_actual = 0
for i in range(numero_de_meses):
    inversion_actual = (inversion_actual + inversion_mensual) * (1 + rentabilidad_efectiva_mensual)
    if (i + 1) % 12 == 0:
        print(f"Año {int((i + 1) / 12)}: ${inversion_actual:,.0f}")
print(f"Ganancia total: ${inversion_actual - inversion_total:,.0f}")
print(f"Porcentaje de ganancia: {(inversion_actual - inversion_total) / inversion_total * 100:.0f}%")
print(f"Tasa de retorno efectiva anual: {((inversion_actual / inversion_total) ** (1 / numero_de_anios) - 1) * 100:.0f}%")

print("\n" + "-" * 50)
print("Si se invierte una sola vez")
print("-" * 50)
print(f"Inversión total: ${inversion_total:,.0f}")
inversion_actual2 = inversion_total
for i in range(numero_de_anios):
    inversion_actual2 = inversion_actual2 * (1 + rentabilidad_efectiva_anual)
    if (i + 1) % 12 == 0:
        print(f"Año {int((i + 1) / 12)}: ${inversion_actual2:,.0f}")
print(f'Ganancia total: ${inversion_actual2 - inversion_total:,.0f}')
print(f'Porcentaje de ganancia: {(inversion_actual2 - inversion_total) / inversion_total * 100:.0f}%')
```
